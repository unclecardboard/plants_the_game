```python
# Clone your repository (with safe directory handling)
import os

# Always start from /content to avoid nested directories
%cd /content

# Remove old clone if it exists
if os.path.exists('plants_the_game'):
  print("🗑️ Removing old clone...")
  !rm -rf plants_the_game

# Clone fresh
print("📥 Cloning repository...")
!git clone -q https://github.com/unclecardboard/plants_the_game.git

# Change into the directory
%cd plants_the_game

print("✅ Repository cloned")
print(f"📁 Working directory: {os.getcwd()}")
```

```python
# Verify assets are present
import os
image_count = len([f for f in os.listdir('assets/images') if f.endswith('.png')])
audio_count = len([f for f in os.listdir('assets/audio') if f.endswith('.mp3')])
print(f"✅ Found {image_count} plant images")
print(f"✅ Found {audio_count} audio files")
print(f"✅ main.py exists: {os.path.exists('main.py')}")
print(f"✅ buildozer.spec exists: {os.path.exists('buildozer.spec')}")
```

```python
# Build the APK (this takes ~30-40 minutes)
# The -v flag gives verbose output so we can see progress
print("🔨 Starting APK build...")
print("⏱️ This will take 30-40 minutes. You can minimize this and come back later.")
print("")
!buildozer -v android debug
print("\n\n✅ APK BUILD COMPLETE!")
print("📱 Find your APK in the 'bin' folder")
```

```python
# List and display the APK
import os
import glob

apk_files = glob.glob('bin/*.apk')
if apk_files:
  print("\n" + "="*60)
  print("🎉 SUCCESS! Your APK is ready:")
  print("="*60)
  for apk in apk_files:
    size_mb = os.path.getsize(apk) / (1024*1024)
    print(f"\n📱 {os.path.basename(apk)}")
    print(f"   Size: {size_mb:.1f} MB")
    print(f"   Path: {apk}")
  print("\n" + "="*60)
  print("📥 TO DOWNLOAD:")
  print("1. Click the 📁 Files icon on the left")
  print("2. Navigate to plants_the_game/bin/")
  print("3. Right-click the .apk file")
  print("4. Select 'Download'")
  print("="*60)
else:
  print("❌ No APK found. Check the build output above for errors.")
```
